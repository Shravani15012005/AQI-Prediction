Filling the null values using mean

```python
pmean=data["PM2.5"].mean()
data["PM2.5"].fillna(pmean,inplace=True)
```

```python
pmmean=data["PM10"].mean()
data["PM10"].fillna(pmmean,inplace=True)
```

```python
nmean=data["NO"].mean()
data["NO"].fillna(nmean,inplace=True)
```

```python
nomean=data["NO2"].mean()
data["NO2"].fillna(nomean,inplace=True)
```

```python
noxmean=data["NOx"].mean()
data["NOx"].fillna(noxmean,inplace=True)
```

```python
nhmean=data["NH3"].mean()
data["NH3"].fillna(nhmean,inplace=True)
```

```python
cmean=data["CO"].mean()
data["CO"].fillna(cmean,inplace=True)
```

```python
smean=data["SO2"].mean()
data["SO2"].fillna(smean,inplace=True)
```

```python
omean=data["O3"].mean()
data["O3"].fillna(omean,inplace=True)
```

```python
bmean=data["Benzene"].mean()
data["Benzene"].fillna(bmean,inplace=True)
```

```python
tmean=data["Toluene"].mean()
data["Toluene"].fillna(tmean,inplace=True)
```

```python
xmean=data["Xylene"].mean()
data["Xylene"].fillna(xmean,inplace=True)
```

```python
amean=data["AQI"].mean()
data["AQI"].fillna(amean,inplace=True)
```

```python
data.isnull().sum()
```

Visualizing various Components in dataset

```python
plt.scatter(data.CO,data.AQI,color='red',marker='*')
plt.plot(data.NO,data.AQI,)
plt.plot
```

```python
city_aqi = data.groupby('City')['AQI'].mean().dropna().sort_values(ascending=False).head(10)
city_aqi.plot(kind='bar', color='orange')
plt.xlabel('City')
plt.ylabel('Average AQI')
plt.title('Top 10 Cities by Average AQI')
plt.show()
```

```python
data['AQI_Bucket'].value_counts().plot(kind='bar', color='purple')
plt.xlabel('AQI Category')
plt.ylabel('Number of Records')
plt.title('Count of AQI Categories')
plt.show()
```
